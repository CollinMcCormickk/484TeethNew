```python
from pathlib import Path
import random
import json
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import cv2


RAD_DIR = Path("Radiographs")
SEG_DIR = Path("Segmentation/teeth_mask")
OUTPUT_SPLIT = Path("data_split.json")
TEST_SIZE = 0.2
SEED = 42


def stem_without_all_suffixes(path: Path) -> str:
    name = path.name
    while True:
        stem = Path(name).stem
        if stem == name:
            return stem
        name = stem


def load_grayscale(path: Path) -> np.ndarray:
    suffix = path.suffix.lower()

    if suffix in [".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff"]:
        img = Image.open(path).convert("L")
        return np.array(img)

    if suffix == ".npy":
        arr = np.load(path)
        if arr.ndim == 3:
            arr = arr[..., 0]
        return arr.astype(np.float32)

    raise ValueError(f"Unsupported image format: {path}")


def basic_threshold(image: np.ndarray) -> np.ndarray:
    img = image.astype(np.float32)
    img = img - img.min()
    if img.max() > 0:
        img = img / img.max()
    img8 = (img * 255).astype(np.uint8)

    if cv2 is not None:
        _, thresh = cv2.threshold(img8, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    else:
        t = img8.mean()
        thresh = (img8 > t).astype(np.uint8) * 255

    return thresh


def build_pairs(rad_dir: Path, seg_dir: Path):
    rad_files = [p for p in rad_dir.iterdir() if p.is_file()]
    seg_files = [p for p in seg_dir.iterdir() if p.is_file()]

    rad_map = {stem_without_all_suffixes(p): p for p in rad_files}
    seg_map = {stem_without_all_suffixes(p): p for p in seg_files}

    common_keys = sorted(set(rad_map) & set(seg_map))
    if not common_keys:
        raise RuntimeError("No matching radiograph/segmentation pairs found by basename.")

    pairs = [(rad_map[k], seg_map[k]) for k in common_keys]
    return pairs


def split_pairs(pairs, test_size=0.2, seed=42):
    rng = random.Random(seed)
    pairs = pairs[:]
    rng.shuffle(pairs)

    n_test = max(1, int(len(pairs) * test_size))
    test_pairs = pairs[:n_test]
    train_pairs = pairs[n_test:]
    return train_pairs, test_pairs


def save_split(train_pairs, test_pairs, out_path: Path):
    split_dict = {
        "train": [
            {"radiograph": str(r), "segmentation": str(s)}
            for r, s in train_pairs
        ],
        "test": [
            {"radiograph": str(r), "segmentation": str(s)}
            for r, s in test_pairs
        ],
    }

    with open(out_path, "w") as f:
        json.dump(split_dict, f, indent=2)


def show_example(test_pairs):
    rad_path, seg_path = test_pairs[0]

    radiograph = load_grayscale(rad_path)
    segmentation = load_grayscale(seg_path)
    thresholded = basic_threshold(radiograph)
    print(radiograph.shape)
    print(segmentation.shape)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(radiograph, cmap="gray")
    axes[0].set_title("Radiograph")
    axes[0].axis("off")

    axes[1].imshow(thresholded, cmap="gray")
    axes[1].set_title("Thresholded")
    axes[1].axis("off")

    axes[2].imshow(segmentation, cmap="gray")
    axes[2].set_title("True Segmentation")
    axes[2].axis("off")

    plt.tight_layout()
    plt.show()



if not RAD_DIR.exists():
    raise FileNotFoundError(f"Missing folder: {RAD_DIR.resolve()}")
if not SEG_DIR.exists():
    raise FileNotFoundError(f"Missing folder: {SEG_DIR.resolve()}")

pairs = build_pairs(RAD_DIR, SEG_DIR)
train_pairs, test_pairs = split_pairs(pairs, test_size=TEST_SIZE, seed=SEED)

print(f"Total pairs: {len(pairs)}")
print(f"Train pairs: {len(train_pairs)}")
print(f"Test pairs: {len(test_pairs)}")
print(f"Saved split file to: {OUTPUT_SPLIT.resolve()}")

show_example(test_pairs)

```

```python
def dice_score(preds, targets, eps=1e-7):
    intersection = (preds * targets).sum()
    union = preds.sum() + targets.sum()
    return ((2 * intersection + eps) / (union + eps)).item()
```

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from PIL import Image

radiograph = np.array(Image.open("Radiographs/9.JPG").convert("L"))
segmentation = np.array(Image.open("Segmentation/teeth_mask/9.jpg").convert("L"))

@interact(threshold=widgets.IntSlider(value=128, min=0, max=255, step=1, continuous_update=False))
def show_threshold(threshold):
    binary = (radiograph > threshold).astype(np.uint8) * 255

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(radiograph, cmap="gray")
    axes[0].set_title("Radiograph")
    axes[0].axis("off")

    axes[1].imshow(binary, cmap="gray", vmin=0, vmax=255)
    axes[1].set_title(f"Threshold = {threshold}")
    axes[1].axis("off")

    axes[2].imshow(segmentation, cmap="gray")
    axes[2].set_title("True Segmentation")
    axes[2].axis("off")

    plt.suptitle(f"Dice: {dice_score(binary,segmentation)}")

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.filters import gaussian
from skimage.segmentation import active_contour

# Load grayscale dental image
img = (mag> 50).astype(np.uint8) * 255
# Contrast enhancement
img_eq = cv2.equalizeHist(img)

# Smooth to reduce noise before snake
img_smooth = gaussian(img_eq, sigma=2, preserve_range=False)

# Initial circular contour around one tooth
h, w = img.shape
cx, cy = w // 2, h // 2          # replace with rough tooth center
r = min(h, w)           # replace with rough tooth radius

s = np.linspace(0, 2*np.pi, 250)
init = np.array([cy + r*np.sin(s), cx + r*np.cos(s)]).T

# Run active contour
snake = active_contour(
    img_smooth,
    init,
    alpha=0.05,     # tension
    beta=0.005,       # smoothness
    gamma=.01,    # step size
    w_edge=1.0,
    w_line=0.0
)

# Visualize
fig, ax = plt.subplots(figsize=(7, 7))
ax.imshow(img, cmap="gray")
ax.plot(init[:, 1], init[:, 0], "--r", lw=2, label="Init")
ax.plot(snake[:, 1], snake[:, 0], "-b", lw=2, label="Snake")
ax.set_title("Tooth segmentation with active contour")
ax.axis("off")
ax.legend()
plt.show()
```

```python
import json
import random
from pathlib import Path

import numpy as np
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

from UnetModel import UNet


SEED = 42
IMAGE_DIR = Path("Radiographs")
MASK_DIR = Path("Segmentation/teeth_mask")
SPLIT_JSON = Path("data_split.json")

IMAGE_SIZE = (256, 512)
BATCH_SIZE = 8
NUM_EPOCHS = 25
LEARNING_RATE = 1e-4
NUM_WORKERS = 2
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
CHECKPOINT_PATH = "best_unet_teeth.pt"


def set_seed(seed: int = SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


class TeethDataset(Dataset):
    def __init__(self, image_paths, mask_paths, image_size=IMAGE_SIZE):
        self.image_paths = image_paths
        self.mask_paths = mask_paths
        self.image_transform = transforms.Compose([
            transforms.Resize(image_size),
            transforms.ToTensor(),
        ])
        self.mask_transform = transforms.Compose([
            transforms.Resize(image_size, interpolation=transforms.InterpolationMode.NEAREST),
            transforms.ToTensor(),
        ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert("L")
        mask = Image.open(self.mask_paths[idx]).convert("L")

        image = self.image_transform(image)
        mask = self.mask_transform(mask)
        mask = (mask > 0.5).float()

        return image, mask


def resolve_path(base_dir: Path, name: str):
    path = Path(name)
    if path.exists():
        return path
    candidate = base_dir / name
    if candidate.exists():
        return candidate
    raise FileNotFoundError(f"Could not find file: {name}")


def load_split_paths(split_json_path: Path, image_dir: Path, mask_dir: Path):
    if not split_json_path.exists():
        raise FileNotFoundError(f"Missing split file: {split_json_path}")

    with open(split_json_path, "r") as f:
        split_data = json.load(f)

    if "train" not in split_data or "test" not in split_data:
        raise KeyError("data_split.json must contain 'train' and 'test' keys")

    train_entries = split_data["train"]
    test_entries = split_data["test"]

    train_imgs, train_masks = parse_entries(train_entries, image_dir, mask_dir)
    test_imgs, test_masks = parse_entries(test_entries, image_dir, mask_dir)

    return train_imgs, train_masks, test_imgs, test_masks


def parse_entries(entries, image_dir: Path, mask_dir: Path):
    image_paths = []
    mask_paths = []

    for entry in entries:
        if isinstance(entry, dict):
            image_name = entry.get("radiograph") 
            mask_name = entry.get("segmentation")
            if image_name is None or mask_name is None:
                raise ValueError(
                    "Each dict entry in data_split.json must contain image/mask fields, "
                    "e.g. {'image': 'x.png', 'mask': 'x.png'}"
                )
        elif isinstance(entry, str):
            image_name = entry
            mask_name = entry
        else:
            raise ValueError("Unsupported entry format in data_split.json")

        image_paths.append(resolve_path(image_dir, image_name))
        mask_paths.append(resolve_path(mask_dir, mask_name))

    return image_paths, mask_paths


def dice_score_from_logits(logits, targets, eps=1e-7):
    probs = torch.sigmoid(logits)
    preds = (probs > 0.5).float()
    intersection = (preds * targets).sum(dim=(1, 2, 3))
    union = preds.sum(dim=(1, 2, 3)) + targets.sum(dim=(1, 2, 3))
    dice = (2 * intersection + eps) / (union + eps)
    return dice.mean().item()


def evaluate(model, loader, criterion, device):
    model.eval()
    loss_meter = 0.0
    dice_meter = 0.0

    with torch.no_grad():
        for images, masks in loader:
            images = images.to(device)
            masks = masks.to(device)

            logits = model(images)
            loss = criterion(logits, masks)

            loss_meter += loss.item() * images.size(0)
            dice_meter += dice_score_from_logits(logits, masks) * images.size(0)

    n = len(loader.dataset)
    return loss_meter / n, dice_meter / n


def main():
    set_seed(SEED)

    if not IMAGE_DIR.exists():
        raise FileNotFoundError(f"Missing image directory: {IMAGE_DIR}")
    if not MASK_DIR.exists():
        raise FileNotFoundError(f"Missing mask directory: {MASK_DIR}")

    train_imgs, train_masks, test_imgs, test_masks = load_split_paths(
        SPLIT_JSON, IMAGE_DIR, MASK_DIR
    )

    train_dataset = TeethDataset(train_imgs, train_masks)
    test_dataset = TeethDataset(test_imgs, test_masks)

    generator = torch.Generator()
    generator.manual_seed(SEED)

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
        generator=generator,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
    )

    model = UNet(in_channels=1, out_channels=1).to(DEVICE)
    criterion = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)

    best_dice = -1.0

    print(f"Using device: {DEVICE}")
    print(f"Train: {len(train_dataset)} | Test: {len(test_dataset)}")

    for epoch in range(NUM_EPOCHS):
        model.train()
        running_loss = 0.0

        for images, masks in train_loader:
            images = images.to(DEVICE)
            masks = masks.to(DEVICE)

            optimizer.zero_grad()
            logits = model(images)
            loss = criterion(logits, masks)
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)

        train_loss = running_loss / len(train_loader.dataset)
        test_loss, test_dice = evaluate(model, test_loader, criterion, DEVICE)

        if test_dice > best_dice:
            best_dice = test_dice
            torch.save({
                "epoch": epoch + 1,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "test_dice": test_dice,
                "seed": SEED,
            }, CHECKPOINT_PATH)

        print(
            f"Epoch [{epoch + 1}/{NUM_EPOCHS}] "
            f"Train Loss: {train_loss:.4f} | "
            f"Test Loss: {test_loss:.4f} | "
            f"Test Dice: {test_dice:.4f}"
        )

    print(f"Best model saved to {CHECKPOINT_PATH} with Dice = {best_dice:.4f}")


if __name__ == "__main__":
    main()
```
